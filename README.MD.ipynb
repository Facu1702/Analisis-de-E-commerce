## 🧠 Descripción del Proyecto

Este proyecto consiste en un análisis comparativo de las ventas realizadas por cuatro tiendas diferentes. El objetivo principal es identificar cuál de ellas presenta un desempeño inferior y explorar posibles causas mediante visualizaciones y métricas clave.

Este trabajo fue realizado con el fin de demostrar mis habilidades analíticas, técnicas y de visualización de datos, como parte de mi desarrollo profesional en el área de análisis de datos.

---

## 📌 Tecnologías y Herramientas Utilizadas

- **Python 3**
- **Jupyter Notebook** (Visual Studio Code)
- **Librerías:**
  - `pandas`
  - `seaborn`
  - `matplotlib`

---

## 📈 Objetivos del Análisis

- Explorar los datos de ventas por tienda.
- Visualizar las tendencias y diferencias entre tiendas.
- Determinar qué tienda tiene el peor desempeño.
- Proponer recomendaciones basadas en datos.
